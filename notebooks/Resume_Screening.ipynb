```python
#import path for handling file and folder paths
#pdfplumber to extract etxt from pdf
from pathlib import Path
import pdfplumber
```

```python
#to go to notebook folder,define loaction of data folder,resumes folder,job descriptionfile
#displaying all paths to verify it
PROJECT_ROOT = Path("..")

DATA_DIR = PROJECT_ROOT / "data"
RESUME_DIR = DATA_DIR / "resumes"
JOB_DESCRIPTION_PATH = DATA_DIR / "job_description.txt"

print("Project root:", PROJECT_ROOT.resolve())
print("Data directory:", DATA_DIR.resolve())
print("Resume directory:", RESUME_DIR.resolve())
print("Job description:", JOB_DESCRIPTION_PATH.resolve())
```

```python
#to verify that file exists or not
print("Data directory exists:", DATA_DIR.exists())
print("Resume directory exists:", RESUME_DIR.exists())
print("Job description exists:", JOB_DESCRIPTION_PATH.exists())
```

```python
#open job_description text file in read mode,read into python string,display loaded job description
with open(JOB_DESCRIPTION_PATH, "r", encoding="utf-8") as file:
    job_description = file.read()

print(job_description)
```

```python
#to check data type,char,words length
print("Type:", type(job_description))
print("Characters:", len(job_description))
print("Words:", len(job_description.split()))
```

```python
#Find every PDF file inside the resumes directory
# sorted() keeps the order consistent
resume_files = sorted(RESUME_DIR.glob("*.pdf"))

print("Number of resumes found:", len(resume_files))

for resume_file in resume_files:
    print("-", resume_file.name)
```

```python
def extract_text_from_pdf(pdf_path):
    """
    Extract text from all pages of a PDF document.

    Parameters
    ----------
    pdf_path : Path
        Path to the PDF file.

    Returns
    -------
    str
        Extracted text from the PDF.
    """
# Store extracted text from each page
    extracted_text = []
# Open the PDF file
    with pdfplumber.open(pdf_path) as pdf:
# Process every page in the PDF
        for page in pdf.pages:
  # Extract text from the current page
            page_text = page.extract_text()
 # Only store text if something was actually extracted
            if page_text:
                extracted_text.append(page_text)
 # Combine text from all pages into one string
    return "\n".join(extracted_text)
```

```python
## Select the first resume from the list
first_resume = resume_files[0]
## Extract text
resume_text = extract_text_from_pdf(first_resume)
# Display first 3000 characters
print(resume_text[:3000])
```

```python
#display name of resume,characters,words
print("Resume:", first_resume.name)
print("Characters:", len(resume_text))
print("Words:", len(resume_text.split()))
```

```python
if resume_text.strip():
    print("Text extraction successful.")
else:
    print("WARNING: No text was extracted from this PDF.")
```

```python
#extract all resumes
resumes = {}

for resume_file in resume_files:

    try:
        text = extract_text_from_pdf(resume_file)

        if text.strip():
            resumes[resume_file.name] = text
        else:
            print(f"WARNING: No text extracted from {resume_file.name}")

    except Exception as e:
        print(f"ERROR processing {resume_file.name}: {e}")
```

```python
print("Successfully processed resumes:", len(resumes))
```
